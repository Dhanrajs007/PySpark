## **Converting RDD (Resilient Distributed Datasets) into Dataframes**
createDataFrame / toDF function

```python
# Creating RDD and converting it into dataset (Object oriented way)

# importing necessary libraries
from pyspark.sql import SparkSession

# function to create new SparkSession
def create_session():
  spk = SparkSession.builder \
      .appName("Corona_cases_statewise.com") \
      .getOrCreate()
  return spk

# function to create RDD
def create_RDD(sc_obj,data):
  df = sc.parallelize(data)
  return df

# function to convert RDD to dataframe
def RDD_to_df(spark,df,schema):

  # converting RDD to df using createDataframe()
  # in which we are passing RDD and schema of df
  df1 = spark.createDataFrame(df,schema)
  return df1

if __name__ == "__main__":

  input_data = [("Uttar Pradesh",122000,89600,12238),
          ("Maharashtra",454000,380000,67985),
          ("Tamil Nadu",115000,102000,13933),
          ("Karnataka",147000,111000,15306),
          ("Kerala",153000,124000,5259)]

  # calling function to create SparkSession
  spark = create_session()

  # creating spark context object
  sc = spark.sparkContext

  # calling function to create RDD
  rd_df = create_RDD(sc,input_data)

  schema_lst = ["State","Cases","Recovered","Deaths"]

  # calling function to convert RDD to dataframe
  converted_df = RDD_to_df(spark,rd_df,schema_lst)

  # visualizing the schema and dataframe
  print(type(rd_df))
  converted_df.printSchema()
  converted_df.show()
```

## **Creating dataframe from List of CustomTypes(datatypes)**
We can create custome types using customType class as shown below and the row class to create row objects to be created in the list of objects

```python
## Approach 1 - using multiple lists of custom objects
# Importing required modules
from pyspark.sql import Row
from pyspark.sql import SparkSession

# Create a SparkSession
spark = SparkSession.builder.appName("MyApp").getOrCreate()
# Define a custom class to represent a row in the dataframe
class CustomType:
    def __init__(self, name,
                 age, salary):
        self.name = name
        self.age = age
        self.salary = salary

# Create a list of CustomType objects
data = [CustomType("John", 30, 5000),
        CustomType("Mary", 25, 6000),
        CustomType("Mike", 35, 7000)]

# Convert the list of CustomType
# objects to a list of Row objects
rows = [Row(name=d.name,
            age=d.age,
            salary=d.salary) for d in data]

# Create a dataframe from the list of Row objects
df = spark.createDataFrame(rows)

# Show the dataframe
df.show()
```

```python
## Approch 2 - Creating custom types using RDD

from pyspark.sql import Row
from pyspark.sql import SparkSession
# Create a SparkSession
spark = SparkSession.builder.appName("Metadata").getOrCreate()
# Define a custom class to represent a row in the dataframe
class CustomType:
    def __init__(self, name, age, salary):
        self.name = name
        self.age = age
        self.salary = salary

# Create a list of CustomType objects
data = [CustomType("John", 30, 5000),
        CustomType("Mary", 25, 6000),
        CustomType("Mike", 35, 7000)]

rdd = spark.sparkContext.parallelize(data)

# Create a dataframe from the rdd
df = spark.createDataFrame(rdd)

# Show the dataframe
df.show()
```

```python
## Approach 3 - Creating the schema using the struct

# Importing required modules
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, IntegerType
from pyspark.sql import Row

# Create a SparkSession
spark = SparkSession.builder.appName("Myapp").getOrCreate()

# step 1: Define the schema for the dataframe
schema = StructType([
    StructField("name",
                StringType(), True),
    StructField("age",
                IntegerType(), True),
    StructField("salary",
                IntegerType(), True)
])

# step 2: Create a list of custom objects
data = [{"name": "John",
         "age": 30, "salary": 5000},
        {"name": "Mary",
         "age": 25, "salary": 6000},
        {"name": "Mike",
         "age": 35, "salary": 7000}]

# step 3: Create the dataframe
df = spark.createDataFrame(data, schema)

# step 4: Show the dataframe
df.show()
```

## **Basic Transformations**
Perform transformations like joins, filters and mappings on your datasets.

```python
import pyspark
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('sparkdf').getOrCreate() # Init SparkSession

d = [["1", "sravan", "company 1"],
        ["2", "ojaswi", "company 1"],
        ["3", "rohith", "company 2"],
        ["4", "sridevi", "company 1"],
        ["5", "bobby", "company 1"]]

cols = ['ID', 'Name', 'Company']
df1 = spark.createDataFrame(d, cols)
df1.show()
```

```python
import pyspark
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('sparkdf').getOrCreate() # Init SparkSession

d1 = [["1", "45000", "IT"],
         ["2", "145000", "Manager"],
         ["6", "45000", "HR"],
         ["5", "34000", "Sales"]]

cols = ['ID', 'salary', 'department']
df2 = spark.createDataFrame(d1, cols)

df2.show()
```

```python
#innner join

df1.join(df2, df1.ID == df2.ID,'inner').show()
```

```python
#Full outer join

# using full keyword
df1.join(df2, df1.ID == df2.ID, 'full').show()

# using fullouter keyword
df1.join(df2, df1.ID == df2.ID, 'fullouter').show()

# using outer keyword
df1.join(df2, df1.ID == df2.ID, 'outer').show()
```

```python
# left join / left outer

# using left keyword
df1.join(df2, df1.ID == df2.ID, 'left').show()

# using left outer keyword
df1.join(df2, df1.ID == df2.ID, 'leftouter').show()
```

```python
# Right join / rightouter join

# using right keyword
df1.join(df2, df1.ID == df2.ID, 'right').show()

# using rightouter keyword
df1.join(df2, df1.ID == df2.ID, 'rightouter').show()
```

```python
# left semi - only matching from left

df1.join(df2, df1.ID == df2.ID, 'leftsemi').show()
```

```python
# left anti - not matching from left

df1.join(df2, df1.ID == df2.ID, 'leftanti').show()
```

```python
# Create temp views
df1.createOrReplaceTempView("student")
df2.createOrReplaceTempView("department")

# SQL join on ID
spark.sql("""
    SELECT * FROM student
    JOIN department
    ON student.ID = department.ID
""").show()
```

```python
# Adding the costant column

# Using lit or sql query

# Create a spark session
from pyspark.sql import SparkSession
from pyspark.sql.functions import lit
spark = SparkSession.builder.appName('SparkExamples').getOrCreate()

# Create a spark dataframe
columns = ["Name", "Course_Name",
           "Months",
           "Course_Fees", "Discount",
           "Start_Date", "Payment_Done"]
data = [
    ("Amit Pathak", "Python", 3,
     10000, 1000, "02-07-2021", True),
    ("Shikhar Mishra", "Soft skills",
     2, 8000, 800, "07-10-2021", False),
    ("Shivani Suvarna", "Accounting", 6,
     15000, 1500, "20-08-2021", True),
    ("Pooja Jain", "Data Science", 12,
     60000, 900, "02-12-2021", False),
]
df = spark.createDataFrame(data).toDF(*columns)

# View the dataframe
df.show()

# Adding a constant column using lit
df.withColumn("Status", lit(2)).show()  # constant value

df1=df.withColumn("Status", lit(2))
# Importing required functions
from pyspark.sql.functions import when, lit, col
df1.withColumn("Course_Status", when(col("Course_Fees") >= 15000,lit("High")) \
              .otherwise(lit("Moderate"))).show()                 # values based on another column
```
